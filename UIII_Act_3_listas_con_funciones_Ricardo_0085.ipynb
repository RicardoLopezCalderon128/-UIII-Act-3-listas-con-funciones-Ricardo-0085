```python
# Diccionario global para almacenar los datos de los diseñadores [6, 7]
# La clave será el ID y el valor un diccionario con los detalles [8]
disenadores = {}

def agregar_disenador():
    """Solicita datos y los guarda en el diccionario [8, 9]."""
    id_dis = input("Ingresa el ID único del diseñador: ")
    if id_dis in disenadores:
        print("Error: Este ID ya existe.")
    else:
        nombre = input("Nombre completo: ")
        especialidad = input("Especialidad (Gráfico, Web, Moda): ")
        # Se guarda la información en un diccionario anidado [10]
        disenadores[id_dis] = {"nombre": nombre, "especialidad": especialidad}
        print(f"Diseñador {nombre} agregado con éxito.")

def listar_disenadores():
    """Recorre y muestra todos los registros almacenados [11, 12]."""
    if not disenadores:
        print("La lista está vacía.")
    else:
        print("\n--- Lista de Diseñadores ---")
        for id_dis, datos in disenadores.items(): # Obtiene pares clave-valor [10]
            print(f"ID: {id_dis} | Nombre: {datos['nombre']} | Especialidad: {datos['especialidad']}")

def buscar_disenador():
    """Busca un diseñador por su ID [13, 14]."""
    id_buscado = input("Ingresa el ID que deseas buscar: ")
    if id_buscado in disenadores: # Comprueba pertenencia [15, 16]
        return id_buscado
    else:
        print("Diseñador no encontrado.")
        return None

def leer_disenador():
    """Muestra la información detallada de un diseñador específico [17, 18]."""
    id_dis = buscar_disenador()
    if id_dis:
        datos = disenadores[id_dis]
        print(f"\nFicha del Diseñador:\nID: {id_dis}\nNombre: {datos['nombre']}\nEspecialidad: {datos['especialidad']}")

def modificar_disenador():
    """Actualiza los datos de un registro existente [8, 19]."""
    id_dis = buscar_disenador()
    if id_dis:
        print("Ingresa los nuevos datos (deja en blanco para mantener el actual):")
        nuevo_nombre = input(f"Nuevo nombre [{disenadores[id_dis]['nombre']}]: ")
        nueva_especialidad = input(f"Nueva especialidad [{disenadores[id_dis]['especialidad']}]: ")

        # Actualización condicional de los campos [20]
        if nuevo_nombre:
            disenadores[id_dis]['nombre'] = nuevo_nombre
        if nueva_especialidad:
            disenadores[id_dis]['especialidad'] = nueva_especialidad
        print("Datos modificados correctamente.")

def borrar_disenador():
    """Elimina un diseñador del sistema [21, 22]."""
    id_dis = buscar_disenador()
    if id_dis:
        # Se elimina la entrada del diccionario usando la clave [21]
        eliminado = disenadores.pop(id_dis)
        print(f"El diseñador {eliminado['nombre']} ha sido borrado.")

# Estructura de control principal (Ciclo interactivo) [3, 23]
while True:
    print("\n--- SISTEMA DE GESTIÓN: DISEÑADOR ---")
    print("1. Agregar Diseñador")
    print("2. Listar todos")
    print("3. Leer ficha específica")
    print("4. Buscar (verificar existencia)")
    print("5. Modificar")
    print("6. Borrar")
    print("7. Salir")

    opcion = input("Selecciona una opción (1-7): ") # Entrada de usuario [9]

    if opcion == '1':
        agregar_disenador()
    elif opcion == '2':
        listar_disenadores()
    elif opcion == '3':
        leer_disenador()
    elif opcion == '4':
        if buscar_disenador(): print("El diseñador está registrado.")
    elif opcion == '5':
        modificar_disenador()
    elif opcion == '6':
        borrar_disenador()
    elif opcion == '7':
        print("Saliendo del programa...")
        break # Rompe el ciclo while para finalizar [4, 24]
    else:
        print("Opción no válida, intenta de nuevo.") # Manejo de errores simple [25]
```
